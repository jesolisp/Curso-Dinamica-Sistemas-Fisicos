# Otro tipo de sistemas

## Sistemas caóticos

### Atractor de Lorenz

Propuesto por Edward Lorenz en 1963, este sistema simplifica las ecuaciones de la convección térmica en la atmósfera. Es el ejemplo clásico de un sistema no lineal autónomo de tercer orden que exhibe caos determinista.

$$
 \begin{aligned}
  \dot{x} &= \sigma\left(y - x\right), \\
  \dot{y} &= x\left(\rho - z\right) - y, \\
  \dot{z} &= xy - \beta z,\\
 \end{aligned}
$$

**Variables del sistema:**
*   $x$: Intensidad del movimiento convectivo (velocidad del fluido).
*   $y$: Diferencia de temperatura horizontal entre las corrientes ascendentes y descendentes.
*   $z$: Distorsión del perfil vertical de temperatura respecto a un gradiente lineal.

**Parámetros y valores clásicos:**
*   $\sigma$ (**Número de Prandtl**): Relación entre la viscosidad cinemática y la difusividad térmica ($\sigma = 10$).
*   $\rho$ (**Número de Rayleigh normalizado**): Representa la diferencia de temperatura impulsora entre la parte inferior y superior del fluido ($\rho = 28$ para régimen caótico).
*   $\beta$ (**Factor geométrico**): Relación de aspecto de las celdas de convección ($\beta = \frac{8}{3} \approx 2.667$).

## Sistemas biológicos

### Modelo presa-depredador

El modelo de Lotka-Volterra describe la dinámica de poblaciones ecológicas en las que interactúan un depredador y su presa:

$$
 \begin{aligned}
  \dot{x} &= \alpha x - \beta xy, \\
  \dot{y} &= \delta xy - \gamma y, \\
 \end{aligned}
$$

**Variables del sistema:**
*   $x$: Número de individuos de la población de presas.
*   $y$: Número de individuos de la población de depredadores.

**Parámetros:**
*   $\alpha$: Tasa de crecimiento natural de las presas en ausencia de depredadores.
*   $\beta$: Tasa de mortalidad de las presas debida a la depredación.
*   $\delta$: Tasa de crecimiento de los depredadores basada en la cantidad de presas consumidas.
*   $\gamma$: Tasa de mortalidad natural de los depredadores en ausencia de presas.

### Modelo SEIR

El modelo SEIR es un sistema compartimental utilizado en epidemiología para describir la propagación de enfermedades infecciosas donde existe un período de incubación significativo. Su dinámica se define mediante el siguiente sistema de ecuaciones diferenciales:

$$
 \begin{aligned}
  \dot{S} &= -\beta S I, \\
  \dot{E} &= \beta S I - \sigma E, \\
  \dot{I} &= \sigma E - \gamma I, \\
  \dot{R} &= \gamma I,
 \end{aligned}
$$

**Variables del sistema:**
*   $S$ (**Susceptibles**): Personas sanas que aún no han contraído la enfermedad y pueden infectarse.
*   $E$ (**Expuestos / Latentes**): Personas que han contraído el patógeno pero se encuentran en período de incubación.
*   $I$ (**Infecciosos**): Personas enfermas que pueden transmitir activamente la infección.
*   $R$ (**Recuperados / Removidos**): Personas que se han recuperado (con inmunidad) o han fallecido.

**Parámetros:**
*   $\beta$: Tasa de transmisión de la infección.
*   $\sigma$: Tasa a la cual los individuos expuestos pasan al estado infeccioso (inverso del período de incubación).
*   $\gamma$: Tasa de recuperación de los individuos infecciosos (inverso del período infeccioso).
